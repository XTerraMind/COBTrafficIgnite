```python
def compute_delay_sig(volume, green_time, cycle_length):
    x = volume / (green_time / cycle_length)
    delay = (0.5 * cycle_length * ((1 - (green_time / cycle_length))**2))
    return delay, x


def LOS_from_delay(delay):
    if delay <= 10: return 'A'
    elif delay <= 20: return 'B'
    elif delay <= 35: return 'C'
    elif delay <= 55: return 'D'
    elif delay <= 80: return 'E'
    else: return 'F'
```
